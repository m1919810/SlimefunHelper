```python
import os
import json
import re
```

```python
directory = "workspace"
targe_directory = "new-version"
files = [ i  for i in os.listdir(directory) if i.endswith(".json")]
display(files)
```

```python
for filename in files:
    with open(os.path.join(directory,filename),mode="r") as fp:
        data = json.load(fp)
    display(data)
    val= data["textures"]["layer0"]
    val1 = val.replace("minecraft:item","slimefunhelper:new-version")
    data["textures"]["layer0"] = val1
    with open(os.path.join(targe_directory,filename),mode="w") as fp:
        json.dump(data,fp)
        
```

```python
atlase = []
for filename in files:
    atlase .append({
            "type": "single",
            "resource": "slimefunhelper:gce/"+(filename.replace(".json",""))
        })
with open("create/blocks.json","w") as fp:
    json.dump(atlase,fp)
```
